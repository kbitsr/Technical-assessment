# Ejercicio # 1  
1.1 Se solicita elaborar un reporte que calcule de forma automática la siguiente información:  Equipo, Grupo al que pertenece, Total Puntos Obtenidos en todos los partidos y Sumatoria del Gol Diferencia logrado en cada partido.

```python

conn = sqlite3.connect('test.db')
####PROGRAME SU RESPUESTA DENTRO DEL CONN.EXECUTE ENTRE LAS COMILLAS
cursor = conn.execute('''

WITH partidos_p AS (
    select
        grupo,
        TRIM(SUBSTR(partido, 1, INSTR(partido, ' vs. ') - 1)) AS local,
        TRIM(SUBSTR(partido, INSTR(partido, ' vs. ') + 5)) AS visitante,
        CAST(SUBSTR(resultado, 1, INSTR(resultado, ',') - 1) AS INTEGER) AS golesLocal,
        CAST(SUBSTR(resultado, INSTR(resultado, ',') + 1) AS INTEGER) AS golesVisitante
    FROM Partidos
),

equipos AS (
    select
        local AS equipo,
        grupo,
        golesLocal AS golesFavor,
        golesVisitante AS golesContra,
        CASE
            WHEN golesLocal > golesVisitante THEN 3
            WHEN golesLocal = golesVisitante THEN 1
            ELSE 0
        END AS puntos
    FROM partidos_p

    UNION ALL

    select
        visitante AS equipo,
        grupo,
        golesVisitante AS golesFavor,
        golesLocal AS golesContra,
        CASE
            WHEN golesVisitante > golesLocal THEN 3
            WHEN golesVisitante = golesLocal THEN 1
            ELSE 0
        END AS puntos
    FROM partidos_p
)

select
    equipo,
    grupo,
    SUM(puntos) AS puntos,
    SUM(golesFavor - golesContra) AS gol_diferencia
FROM equipos
GROUP BY equipo, grupo
                       ''')

for row in cursor:
  print(row)
conn.close()
```

# Ejercicio # 2


2.1.   Determinar los 2 equipos con mayor puntaje que pasarán a la siguiente ronda por cada grupo.  En caso de que existan mas de dos equipos con derecho a pasar por tener igual número de puntos, se definirá quien pasa considerando el que tenga el mayor gol diferencia.

2.2.   Agregar un campo indicando en qué ciudad y Fecha/hora que jugarán en la siguiente ronda los equipos clasificados.

```python

conn = sqlite3.connect('test.db')
cursor = conn.execute('''

WITH Ranking AS (
    SELECT
        equipo,
        grupo,
        puntos,
        gol_diferencia,
        ROW_NUMBER() OVER(
            PARTITION BY grupo
            ORDER BY puntos DESC, gol_diferencia DESC
        ) AS pos
    FROM Resultados_Qatar
),

Top2 AS (
    SELECT
        *,
        CAST(pos AS VARCHAR) || '° del grupo ' || REPLACE(grupo, 'GRUPO ', '') AS clave
    FROM Ranking
    WHERE pos <= 2
)
SELECT
    t.grupo,
    t.pos AS posicion,
    t.equipo,
    t.puntos,
    t.gol_diferencia,
    --TRIM(SUBSTR(c.sede, INSTR(c.sede, ',') + 1)) || ' - ' || c.fecha || ' ' || c.hora AS ciudad_fecha_hora
    c.sede AS Ciudad,
    t.clave
FROM Top2 t
LEFT JOIN Clasificados c
    ON c.partido LIKE '%' || t.clave || '%'
ORDER BY t.grupo, t.pos
                       ''')

for row in cursor:
  print(row)
conn.close()
```

```python
conn = sqlite3.connect('test.db')
cursor = conn.execute('''

WITH Quintil AS (
    select
        equipo,
        puntos,
        gol_diferencia,
        NTILE(5) OVER (ORDER BY puntos desc, gol_diferencia desc) AS quintil
    FROM Resultados_Qatar
)
select
    equipo,
    puntos
FROM Quintil
WHERE quintil = 1
ORDER BY puntos desc, gol_diferencia desc
                      ''')
for row in cursor:
  print(row)
conn.close()
```

```python
# 5.2
df_promedios = df.groupBy("centro_titularidad").agg(
    F.avg("direccion_coor_x").alias("prom_x"),
    F.avg("direccion_coor_y").alias("prom_y")
)

df_con_promedios = df.join(df_promedios, on="centro_titularidad", how="inner")

df_con_distancia = df_con_promedios.withColumn(
    "distancia",
    F.sqrt(
        F.pow(F.col("direccion_coor_x") - F.col("prom_x"), 2) +
        F.pow(F.col("direccion_coor_y") - F.col("prom_y"), 2)
    )
)
window_spec = Window.partitionBy("centro_titularidad").orderBy(F.col("distancia").asc())

df_centro_mas_centrico = df_con_distancia \
    .withColumn("ranking", F.row_number().over(window_spec)) \
    .filter(F.col("ranking") == 1)

# Mostrar el resultado final con los campos correctos del JSON
df_centro_mas_centrico.select(
    "centro_titularidad",
    "centro_nombre",
    "municipio_nombre",
    "direccion_coor_x",
    "direccion_coor_y",
    "distancia"
).show(20, truncate=False)
```
